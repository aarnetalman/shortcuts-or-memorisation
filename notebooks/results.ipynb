```python
import pandas as pd
from nlicorrupt import paths

res = paths.results_dir()
df = pd.read_csv(res / 'all_results.csv')
ct = pd.read_csv(res / 'tables' / 'corrupt_test.csv')
df.groupby('family').size()
```

## Reproduction of the 2021 and 2022 numbers

```python
pd.read_csv(res / 'tables' / 'reproduction.csv')
```

## Corrupt-Test Δ per system (corrupted − clean, primary metric)

```python
view = ct[ct.corruption.isin(['drop-NOUN', 'drop-VERB', 'shuffle-n1', 'hyp-only'])]
view.pivot_table(index=['family', 'model', 'condition'], columns=['task', 'corruption'], values='delta').round(1)
```
